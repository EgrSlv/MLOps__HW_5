```python
import pandas as pd
import sqlite3

def get_employees(min_age=30):
    data = {
        'id': [1, 2, 3, 4, 5],
        'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
        'age': [25, 32, 37, 28, 41],
        'salary': [50000, 60000, 75000, 52000, 90000]
    }

    df = pd.DataFrame(data)
    conn = sqlite3.connect(':memory:')
    df.to_sql('employees', conn, index=False)

    query = "SELECT * FROM employees WHERE age > 30"
    result = pd.read_sql(query, conn)

    print(f"Найдено сотрудников старше 30 лет: {len(result)}")
    print(result)

    filtered = pd.read_sql(f"SELECT * FROM employees WHERE age >= {min_age}", conn)
    conn.close()

    return filtered

def _main_():
    result = get_employees()
    print("\nОтфильтровано:", result)

_main_()
```

```python
import marimo as mo

min_age = mo.ui.slider(20, 50, value=30, label="Минимальный возраст")
min_age
```

```python
result = get_employees(min_age.value)
mo.md(f"### Найдено сотрудников: **{len(result)}**")
result
```

```python
if len(result) > 0:
    print(f"Средняя зарплата: {result['salary'].mean():.2f}")
    print(f"Макс зарплата: {result['salary'].max()}")
    print(f"Мин возраст: {result['age'].min()}")
```
